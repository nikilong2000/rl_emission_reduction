```python
# ==============================================================================
# EJECUCIÓN DEL EXPERIMENTO DE COMPARACIÓN
# ==============================================================================
if __name__ == "__main__":
    from transition_function_model import *

    
    # 1. Definición de rutas
    ruta_ICE = "../src/models_markus/ICE_Model_Update_01"
    ruta_PG  = "../src/models_markus/PG_Model_M1.1_without_EM1_Torque"

    print(f"🔵 Cargando módulos desde 'transition_function_model.py'...")

    # 2. Instanciamos el entorno
    try:
        tf_env = setup_transition_function_model(ruta_ICE, ruta_PG)
    except Exception as e:
        print(f"\n❌ Error iniciando el modelo. Verifica las rutas: {e}")
        exit()
    
    # 3. Reset inicial
    tf_env.reset_models()

    # 4. Cabecera de la tabla para TODAS las variables
    print("\n📊 --- RESULTADOS EXPERIMENTO (TensorFlow Original) ---")
    # Ajustamos el ancho de columna para que quepa todo
    header = f"{'Iter':<5} | {'Torque':<12} | {'NO':<12} | {'NO2':<12} | {'CO':<12} | {'CO2':<12}"
    print(header)
    print("-" * len(header))

    # 5. Valores de entrada constantes
    inputs = {
        'Speed_rpm': 2000.0, 
        'm_fuel_mg': 35.0, 
        'T_amb_K': 298.0, 
        'p_amb_bar': 1.013
    }

    sp_rpm = tf.constant(inputs['Speed_rpm'], dtype=tf.float32)
    fuel   = tf.constant(inputs['m_fuel_mg'], dtype=tf.float32)
    t_amb  = tf.constant(inputs['T_amb_K'], dtype=tf.float32)
    p_amb  = tf.constant(inputs['p_amb_bar'], dtype=tf.float32)

    # 6. Bucle de 5 llamadas
    for i in range(5):
        # Llamada al método del código antiguo
        Torque_Nm, NO_out, NO2_out, CO_out, CO2_out = tf_env.predict_ice(
            sp_rpm, fuel, t_amb, p_amb
        )
        
        # Extraer valores (numpy)
        v_torque = Torque_Nm.numpy()
        v_no     = NO_out.numpy()
        v_no2    = NO2_out.numpy()
        v_co     = CO_out.numpy()
        v_co2    = CO2_out.numpy()
        
        # Imprimir fila con todas las variables
        print(f"{i+1:<5} | {v_torque:<12.5f} | {v_no:<12.5f} | {v_no2:<12.5f} | {v_co:<12.5f} | {v_co2:<12.5f}")

    print("-" * len(header))
```
