```python
import joblib
import pandas as pd

def summarize_scaler(scaler):
    """
    Returns a DataFrame with information about a fitted MinMaxScaler.
    
    Parameters
    ----------
    scaler : sklearn.preprocessing.MinMaxScaler
        The already fitted scaler.
    
    Returns
    -------
    pd.DataFrame
        Columns: feature, data_min, data_max, feature_range_min,
                 feature_range_max, scale, min.
    """
    # variable names
    features = scaler.feature_names_in_
    
    # decompose feature_range into min and max
    fr_min, fr_max = scaler.feature_range
    
    # build DataFrame
    df = pd.DataFrame({
        'feature':          features,
        'data_min':         scaler.data_min_,
        'data_max':         scaler.data_max_,
        'feature_range_min': fr_min,
        'feature_range_max': fr_max,
        'scale':            scaler.scale_,
        'min':              scaler.min_,
    })
    return df
```

```python
import joblib
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

def summarize_scaler(fitxer: str) -> pd.DataFrame:
    """
    Loads a MinMaxScaler from a file (can be .lib, .joblib, .pkl, etc.)
    and returns a DataFrame with its information.
    
    Parameters
    ----------
    fitxer : str
        Path to the file where the scaler is saved.
    
    Returns
    -------
    pd.DataFrame
        Columns: feature, data_min, data_max, feature_range_min,
                 feature_range_max, scale, min.
    """
    # load the scaler (the extension doesn't matter as long as it's a joblib dump)
    scaler = joblib.load(fitxer)
    
    if not isinstance(scaler, MinMaxScaler):
        raise TypeError(f"The loaded object is not a MinMaxScaler (it is {type(scaler)}).")
    

    # decompose feature_range into min and max
    fr_min, fr_max = scaler.feature_range
    
    # build DataFrame
    df = pd.DataFrame({
        'data_min':          scaler.data_min_,
        'data_max':          scaler.data_max_,
        'feature_range_min': fr_min,
        'feature_range_max': fr_max,
        'scale':             scaler.scale_,
        'min':               scaler.min_,
    })
    return df
```

```python
summarize_scaler("gbm_v1.lib")
```
